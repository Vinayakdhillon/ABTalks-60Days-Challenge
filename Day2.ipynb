```python
# Student Grade Checker Program

try:
    # Taking marks input
    marks = float(input("Enter your marks: "))

    # Validation
    if marks < 0 or marks > 100:
        print("Invalid input! Marks should be between 0 and 100.")

    else:
        # Grade assignment
        if marks >= 90:
            grade = "A"
        elif marks >= 75:
            grade = "B"
        elif marks >= 50:
            grade = "C"
        else:
            grade = "Fail"

        # Display grade
        print("Grade:", grade)

        # Pass/Fail check
        if marks >= 50:
            print("Status: Passed")
        else:
            print("Status: Failed")

except ValueError:
    print("Invalid input! Please enter numeric values only.")
```

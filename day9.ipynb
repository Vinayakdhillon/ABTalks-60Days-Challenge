```python
# Prefix Sum Optimization - Day 9

# Original Array
arr = [2, 4, 1, 6, 3, 5]

# -----------------------------
# Step 1: Build Prefix Sum Array
# -----------------------------
prefix = [0] * len(arr)

prefix[0] = arr[0]

for i in range(1, len(arr)):
    prefix[i] = prefix[i - 1] + arr[i]

print("Original Array:", arr)
print("Prefix Sum Array:", prefix)

# --------------------------------
# Brute Force Range Sum Function
# --------------------------------
def brute_force_sum(left, right):
    total = 0
    for i in range(left, right + 1):
        total += arr[i]
    return total

# --------------------------------
# Optimized Prefix Sum Function
# --------------------------------
def prefix_sum_query(left, right):
    if left == 0:
        return prefix[right]
    return prefix[right] - prefix[left - 1]

# Queries
queries = [(1, 3), (0, 4), (2, 5)]

print("\nRange Sum Queries:")

for left, right in queries:
    brute = brute_force_sum(left, right)
    optimized = prefix_sum_query(left, right)

    print(f"\nQuery ({left}, {right})")
    print("Brute Force Result:", brute)
    print("Optimized Result :", optimized)
```

```python
# Take list input from user
numbers = list(map(int, input("Enter numbers separated by space: ").split()))

# Sum of numbers
total = 0
for num in numbers:
    total += num

# Find maximum
maximum = numbers[0]
for num in numbers:
    if num > maximum:
        maximum = num

# Find minimum
minimum = numbers[0]
for num in numbers:
    if num < minimum:
        minimum = num

# Frequency count using dictionary
frequency = {}

for num in numbers:
    if num in frequency:
        frequency[num] += 1
    else:
        frequency[num] = 1

# Reverse list without using built-in reverse functions
reversed_list = []

for i in range(len(numbers) - 1, -1, -1):
    reversed_list.append(numbers[i])

# Output
print("Sum =", total)
print("Maximum =", maximum)
print("Minimum =", minimum)

print("Frequency of each number:")
for key, value in frequency.items():
    print(key, "->", value)

print("Reversed List =", reversed_list)
```
